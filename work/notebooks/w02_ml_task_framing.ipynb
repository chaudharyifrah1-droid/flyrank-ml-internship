## 5. Why ML beats a fixed rule here

A simple rule like IF age > 180 days THEN refresh fails because traffic decay depends on non-linear interactions across multiple factors: seasonal search trends, search intent shifts, competitive rank updates, and keyword velocity. A fixed rule leads to high false-positive rates (flagging evergreen content) or missed opportunities (missing fast-decaying recent trends). Machine learning dynamically learns patterns across all these features simultaneously.







```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Code demonstration showing rule limitations vs multi-feature logic
def rule_based_check(age_days):
    return 1 if age_days > 180 else 0

# Illustrating how a rule misclassifies an evergreen 200-day-old article
sample_article_age = 200
sample_article_traffic_growing = True

rule_decision = rule_based_check(sample_article_age)
print(f"Fixed Rule Decision (Age > 180): Refresh={rule_decision}")
print("Problem: Article is older than 180 days BUT organic traffic is actively growing.")
print("Conclusion: ML evaluates both age, traffic trends, and competition jointly to avoid this error.")






```
